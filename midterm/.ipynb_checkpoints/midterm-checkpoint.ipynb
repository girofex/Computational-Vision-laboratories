```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from scipy import spatial
from skimage import io, color, img_as_float
from skimage.segmentation import slic, mark_boundaries

frames = []

plt.figure(figsize=(16, 30))

for t in range(30):
    frames.append(io.imread(f'imgs/{t+1}.jpg'))
    
    plt.subplot(10, 3, t+1)
    plt.imshow(frames[t], cmap='gray')
    plt.title(f'Frame {t+1}')
    plt.axis('off')
```

<h3>Functions</h3>

<h4><tt>compute_centroids()</tt></h4>

To obtain the centroids of the superpixels given in input, I compute the average row and column coordinates of their pixels.

```python
def compute_centroids(segment):
    num = segment.max() + 1
    centroids = np.zeros((num, 2), dtype=int)

    for label in range(num):
        r, c = np.nonzero(segment == label)

        if r.size == 0:
            continue

        centroids[label] = [r.mean(), c.mean()]

    return centroids
```

<h4><tt>patch_descriptor()</tt></h4>

The same function used in the lab, where I extract patches of the image around each centroid given by the previous function.

```python
def patch_descriptor(I, cent, size_w):
    n = len(cent)
    hw = int(np.floor(size_w/2))
    I_ext = np.pad(I, hw, 'reflect')
    patches = np.zeros((n, (2*hw+1)**2))

    for i in range(0, n):
        r = cent[i,0]+hw
        c = cent[i,1]+hw
        tmp = I_ext[r-hw:r+hw+1, c-hw:c+hw+1]
        patches[i,:] = tmp.flatten()

    return patches
```

<h4><tt>spectral_matching()</tt></h4>

A variation of the function to detect a change used in the lab.
<br>
I used a percentile of the raw Euclidean distances as a threshold: the superpixels whose appearance difference exceeds it are marked as changed.
<br>
I choose the 90th percentile to avoid capturing the smaller details, therefore having a cleaner output.

```python
def spectral_matching(patches1, patches2, cent1, cent2, sigma, alpha):
    D = spatial.distance.cdist(patches1, patches2, metric='euclidean')
    D_pos = spatial.distance.cdist(cent1, cent2, metric='euclidean')
    E1 = np.exp(-D_pos // (2*sigma*sigma))
    E2 = np.exp(-D // (2*sigma*sigma))
    E = 0.5*(alpha*E1+(1-alpha)*E2)

    argmaxE_h = np.argmax(E, axis=1)
    argmaxE_v = np.argmax(E, axis=0)

    match = np.zeros((len(cent1), 2), dtype=int)

    for i, amx in enumerate(argmaxE_h):
        if argmaxE_v[amx] == i:
            match[i] = cent2[amx]
    
    best_dists = D[np.arange(len(cent1)), argmaxE_h]
    threshold = np.percentile(best_dists, 90)
    changed = best_dists > threshold

    return match, changed
```

<h4><tt>show_match()</tt></h4>

A variation of the one used in the lab, to compute the feature matching only on the changed superpixels given by the previous function.

```python
def show_match(cent, match, img1, img2, changed, superpixels):
    img = np.concatenate([img1, img2], axis=1)
    sup = np.concatenate([superpixels[0], superpixels[1]], axis=1)
    changed_idx = np.where(changed)[0]
    plt.imshow(img, cmap=cm.gist_gray)
    plt.imshow(sup, cmap=cm.gist_gray)

    for i in changed_idx:
        plt.plot([cent[i, 1], match[i, 1] + img1.shape[1]], [cent[i, 0], match[i, 0]], 'y')

    plt.scatter(cent[changed_idx, 1], cent[changed_idx, 0], s=10, c='r')
    plt.scatter(match[changed_idx, 1] + img1.shape[1], match[changed_idx, 0], s=20, c='b')
    plt.axis('off')
```

<h4><tt>compute_segments()</tt></h4>

A new function to perform the main procedure with different numbers of segments.
<br>
Firstly, for each frame I compute the superpixels and their corresponding centroids.
<br>
Then, for each time <tt>t</tt>, I feature match and I create a boolean mask by simply comparing the adjacent frames.

The parameters of the functions used inside of <tt>compute_segments()</tt> are set accordingly:
<ul>
    With <tt>slic()</tt>:
    <li><tt>sigma = 2</tt>, reducing noise but keeping the finer details</li>
    <li><tt>compactness = 10</tt> to mantain spatial coherence but to avoid superpixels that are too square-shaped</li>
</ul>

<ul>
    With <tt>patch_descriptor()</tt>:
    <li><tt>size_w = 25</tt> to capture just enough and to avoid an heavy computation</li>
</ul>

<ul>
    With <tt>spectral_matching()</tt>:
    <li><tt>sigma = 2</tt> like before</li>
    <li><tt>alpha = 0.5</tt> to balance spatial consistency and appearance similarity</li>
</ul>

```python
def compute_segments(frames, n_segments):
    segments, centroids, grays, patches, matches, superpixels = [], [], [], [], [], []
    sigma, compactness, size_w, alpha = 2, 10, 25, 0.5

    for i, f in enumerate(frames):
        seg = slic(f, n_segments, sigma, compactness)
        segments.append(seg)

        cent = compute_centroids(seg)
        centroids.append(cent)

        gray = img_as_float(color.rgb2gray(f))
        grays.append(gray)

        patches.append(patch_descriptor(grays[i], cent, size_w))

    plt.figure(figsize=(16, 30))
    for i, f in enumerate(frames):
        plt.subplot(10, 3, i+1)   
        plt.title(f'Frame {i+1}')
        superpixels.append(mark_boundaries(f, segments[i]))
        plt.imshow(superpixels[i])
        plt.axis('off')
    plt.tight_layout()
    plt.subplots_adjust(top=0.93)
    plt.suptitle("Superpixels")

    masks = [np.zeros_like(seg, dtype=bool) for seg in segments]

    plt.figure(figsize=(16, 35))

    for t in range(len(frames) - 1):
        seg1, seg2 = segments[t], segments[t+1]
        cent1, cent2 = centroids[t], centroids[t+1]
        pat1, pat2 = patches[t], patches[t+1]
        frame1, frame2 = frames[t], frames[t+1]
        sup1, sup2 = superpixels[t], superpixels[t+1]

        match, changed = spectral_matching(pat1, pat2, cent1, cent2, sigma, alpha)
        matches.append((cent1, match, frame1, frame2, changed))

        for lbl in np.where(changed)[0]:
            masks[t][seg1 == lbl] = True

            matched_centroid = match[lbl]
            dists = np.sum((cent2 - matched_centroid) ** 2, axis=1)
            closest_lbl = np.argmin(dists)

            masks[t+1][seg2 == closest_lbl] = True

        plt.subplot(15, 2, t+1)
        plt.title(f"Frame {t+1}-{t+2}: {np.sum(changed)} changed")
        show_match(cent1, match, frame1, frame2, changed, [sup1, sup2])

    plt.tight_layout()
    plt.subplots_adjust(top=0.93)
    plt.suptitle('Feature Matching')

    plt.figure(figsize=(16, 30))
    for t in range(len(frames)):
        plt.subplot(10, 3, t+1)
        plt.title(f'Frame {t+1}')
        plt.imshow(masks[t], cmap='gray', alpha=0.8)
        plt.axis('off')
    plt.tight_layout()
    plt.subplots_adjust(top=0.93)
    plt.suptitle("Masks")
```
